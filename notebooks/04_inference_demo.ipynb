# 04 — Inference Demo

Run the pipeline on a single image or a directory of frames.
Useful for quick sanity checks before batch processing.

```python
import sys
sys.path.insert(0, "../src")

import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from pathlib import Path
```

```python
from walker_gait.pipeline import WalkerGaitPipeline
from walker_gait.utils import draw_keypoints_on_image

pipeline = WalkerGaitPipeline(
    pose_model_name="usyd-community/vitpose-plus-large",
    # pose_checkpoint="../checkpoints/best",  # uncomment to use fine-tuned
    device="cuda",
)
```

## Single image inference

```python
# Replace with your own image path
img_path = "../data/frames/sample.png"  # update this
image = Image.open(img_path).convert("RGB")

result = pipeline(image)

print(f"Detected {len(result.keypoints)} person(s)")
for k, name in enumerate(result.keypoint_names):
    if len(result.keypoints) > 0:
        x, y = result.keypoints[0, k]
        score = result.scores[0, k]
        print(f"  {name:>16s}: ({x:7.1f}, {y:7.1f})  score={score:.3f}")
```

```python
if len(result.keypoints) > 0:
    vis = draw_keypoints_on_image(
        np.array(image), result.keypoints, result.scores
    )
    plt.figure(figsize=(10, 8))
    plt.imshow(vis)
    plt.axis("off")
    plt.title("Walker-Gait: 12 Lower-Body Keypoints")
    plt.show()
```
